```python
#feature selection
from sklearn.model_selection import train_test_split #splitting the data into training and tesing 
from sklearn.preprocessing import StandardScaler #for scaling data
from sklearn.linear_model import LogisticRegression #for the modeling
from sklearn.metrics import classification_report, confusion_matrix #this is for the model evaluation
from sklearn.pipeline import Pipeline #for training the model and doing the transformation operations together
from sklearn.compose import ColumnTransformer #data transformation
from sklearn.preprocessing import OneHotEncoder #data transformation
```

```python
df.head()
```

```python
df_model = df.drop(["nameOrig", "nameDest", "isFlaggedFraud"], axis =1)
```

```python
df_model.head()
```

```python
categorical =["type"]
numeric =["amount","oldbalanceOrg","newbalanceOrig","oldbalanceDest","newbalanceDest"]
```

```python
y = df_model["isFraud"]
X = df_model.drop("isFraud", axis=1)
```

```python
X_train, X_test,y_train, y_test = train_test_split(X,y, test_size=0.3, stratify=y)
```

```python
preprocessor = ColumnTransformer(
    transformers= [
        ("num", StandardScaler(), numeric),
        ("cat", OneHotEncoder(drop="first"), categorical)
    ],
    remainder ="drop"
)
```

```python
#creating the model pipeline
pipeline = Pipeline([
    ("prep", preprocessor),
    ("clf", LogisticRegression(class_weight="balance", max_iter=1000))
]
)
```

```python
#training the model
pipeline = Pipeline([
    ("prep", preprocessor),
    ("clf",LogisticRegression(class_weight="balanced", max_iter=1000))
]
)
```

```python
pipeline.fit(X_train,y_train)
```

```python
 y_pred = pipeline.predict(X_test)
```

```python
print(classification_report(y_test,y_pred))
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(round(pipeline.score(X_test,y_test) *100 ,2,),"%")
```

```python
#exporting the pipeline
import joblib

joblib.dump(pipeline,"Fraud_detection_pipeline.pkl")
```
